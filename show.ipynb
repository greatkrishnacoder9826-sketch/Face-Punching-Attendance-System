```python
import pandas as pd
import numpy as np
import cv2 

date = input("Enter the date: ")

faceDetect = cv2.CascadeClassifier(
    r"C:\MACHINE LEARNING PROJECTS\Face Puching - Open CV\haarcascade_frontalface_default (1).xml"
)

cam = cv2.VideoCapture(0)
rec = cv2.face.LBPHFaceRecognizer_create()
rec.read(r"C:\MACHINE LEARNING PROJECTS\Face Puching - Open CV\trainningData.yml")

font = cv2.FONT_HERSHEY_COMPLEX_SMALL

# Read CSV file
df = pd.read_csv(r"C:\MACHINE LEARNING PROJECTS\Face Puching - Open CV\Book1.csv")

while True:
    ret, img = cam.read()
    if not ret:
        print("Failed to capture image")
        break
        
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # Fixed: BGR2GRAY, not BGRA2GRAY
    faces = faceDetect.detectMultiScale(gray, 1.3, 5)  # Fixed: detectMultiScale (not deteectMultiScale)
    
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x + w, y + h), (0, 0, 255), 2)
        id, conf = rec.predict(gray[y:y + h, x:x + w])
        
        # Fixed: Get name from DataFrame
        name = df[df['Id'] == id]['Names'].values
        if len(name) > 0:
            name = name[0]
            # Update attendance (mark 'P' for present)
            df.loc[df['Id'] == id, date] = 'P'
            # Save to CSV
            df.to_csv(r"C:\MACHINE LEARNING PROJECTS\Face Puching - Open CV\Book1.csv", index=False)
        else:
            name = "Unknown"
        
        cv2.putText(img, str(name), (x, y + h), font, 3, 255)
    
    cv2.imshow('Face', img)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cam.release()
cv2.destroyAllWindows()  # Fixed: destroyAllWindows (not destoryAllWindows)
```
